# AegisAML: 03 - Explainability (SHAP & LIME)

Generating regulator-grade explanations for the ML models.


```python
import shap
import pandas as pd
import numpy as np
```

```python
def generate_shap_values(model, X: pd.DataFrame):
    """Generates SHAP values for tree-based models."""
    explainer = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X)
    return shap_values, explainer
```
