# AegisAML: 04 - MLOps & Registry

MLflow integration, model versioning, and deployment wrappers.


```python
import mlflow
import mlflow.sklearn
import os
```

```python
def save_model_to_mlflow(model, model_name: str):
    """Saves model to MLflow registry."""
    mlflow.set_tracking_uri("sqlite:///mlflow.db")
    with mlflow.start_run():
        mlflow.sklearn.log_model(model, artifact_path="model")
        print(f"Model {model_name} logged to MLflow.")
```
