# 5. Model Explainability (SHAP)

AegisAML is fully transparent. We use SHAP to unbox the XGBoost model so investigators know exactly *why* a SAR is recommended.

```
import shap
import xgboost as xgb
import numpy as np

X = np.random.rand(100, 25)
model = xgb.XGBClassifier().fit(X, np.random.randint(0, 2, 100))

explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X)

print("SHAP values calculated successfully. Ready for Investigator Dashboard rendering.")
```
