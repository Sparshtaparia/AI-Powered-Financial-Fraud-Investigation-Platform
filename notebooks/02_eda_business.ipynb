# 2. Business Insights & Typology EDA

Unlike standard EDA, this notebook explores the raw data strictly through the lens of Financial Crime compliance: Velocity, Cash Ratios, and Geographic Drift.

```
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Simulating Time-Series Velocity
dates = pd.date_range("2026-01-01", periods=30, freq="D")
velocity = np.random.poisson(lam=5000, size=30)

plt.figure(figsize=(12, 4))
plt.plot(dates, velocity, marker="o", color="#39FF14")
plt.title("Transaction Velocity (Daily)")
plt.ylabel("Txns/Day")
plt.grid(alpha=0.2)
plt.style.use("dark_background")
plt.show()
```
