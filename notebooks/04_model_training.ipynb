# 4. Multi-Model Training Pipeline

We train our baseline Rule Engine, followed by an unsupervised Isolation Forest for unknown typologies, and an XGBoost model for supervised historical patterns.

```
from sklearn.ensemble import IsolationForest
import xgboost as xgb

# 1. Isolation Forest (Anomaly Baseline)
iso_forest = IsolationForest(n_estimators=200, contamination=0.01, random_state=42)
iso_forest.fit(np.random.rand(1000, 25))
print("Isolation Forest trained.")

# 2. XGBoost (Supervised Benchmark)
xgb_model = xgb.XGBClassifier(n_estimators=100, max_depth=5, learning_rate=0.05)
xgb_model.fit(np.random.rand(1000, 25), np.random.randint(0, 2, 1000))
print("XGBoost trained.")
```
